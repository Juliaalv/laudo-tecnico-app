```python
import matplotlib.pyplot as plt
import numpy as np

# Labels reorganizados e valores atualizados
labels = [
    'Público (ocupação)',
    'Ventilação',
    'Equipamentos',
    'Insolação total',
    'Iluminação',
    'Infiltração'
]

valores = [
    40286.0,     # Ocupação
    17651.83,    # Ventilação
    3287.5,      # Equipamentos
    93147.53,    # Insolação total (atualizado)
    391.28,      # Iluminação (mesmo valor)
    46251.67     # Infiltração
]

colors = ['#FF6384', '#36A2EB', '#FFCE56', '#4BC0C0', '#9966FF', '#FF9F40']

fig, ax = plt.subplots(figsize=(7, 6))
wedges, _ = ax.pie(valores, colors=colors, startangle=90, wedgeprops=dict(width=0.4))

# Cálculo da posição dos textos sem linhas
total = sum(valores)
for i, (wedge, valor) in enumerate(zip(wedges, valores)):
    ang = (wedge.theta2 + wedge.theta1) / 2
    x = np.cos(np.radians(ang)) * 1.2
    y = np.sin(np.radians(ang)) * 1.2
    percentual = valor / total * 100
    ax.text(x, y, f"{labels[i]}\n{percentual:.1f}%", ha='center', va='center', fontsize=18)


plt.tight_layout()
plt.show()

```
